# Buscar datos

Este cuaderno te permite buscar dentro de todas las declaraciones aduaneras de Uruguay (2016 a hoy) — importaciones, exportaciones y tránsitos — sin necesidad de programar. Cambiá los valores de la celda de "Configuración" y corré el cuaderno de punta a punta.

## Cómo usarlo

1. En la celda de **Configuración**, completá lo que quieras buscar (abajo explicamos cada opción).
2. Corré todas las celdas del cuaderno, en orden.
3. El resultado aparece debajo de la consulta, y también se guarda automáticamente en `outputs/tables/`.

## Qué podés buscar

| Opción | Qué es | Ejemplo |
|---|---|---|
| `query_years` | El año o los años a consultar | `2024` o `[2023, 2024, 2025]` |
| `query_tipo_regimen` *(opcional)* | Limitar a importaciones (`"i"`), exportaciones (`"e"`) o tránsitos (`"t"`) | Dejalo en `None` para incluir todo |
| `query_nombre` *(opcional)* | Nombre (o parte del nombre) de una empresa importadora/exportadora | `"acme"` |
| `query_ncm` *(opcional)* | Código del producto (NCM), completo o solo el comienzo | `"2701"` |
| `query_formato` | En qué formato guardar el resultado | `"csv"` (por defecto), `"xlsx"` o `"parquet"` |

Hace falta completar al menos `query_nombre` o `query_ncm` — si no, la búsqueda no tendría por dónde empezar entre las ~20 millones de filas del dataset.

**Tip:** cuanto más específico el nombre de la empresa, mejor. Buscar solo `"sa"`, por ejemplo, te va a devolver miles de coincidencias distintas.

````python
from uruguay_dua_comex import utils
from uruguay_dua_comex.query import export_query_result, query_dua

EXPORT_DIR = utils.outputs_tables_dir()
````

````python
# --- Configuración ---
query_years = [2022, 2023, 2024, 2025, 2026]
query_tipo_regimen = None  # "i", "e", "t", o None para incluir todo
query_nombre = None
query_ncm = "7801"
query_formato = "csv"  # "csv" (por defecto), "xlsx" o "parquet"

query_result = query_dua(
    years=query_years, tipo_regimen=query_tipo_regimen, nombre=query_nombre, ncm=query_ncm
)
query_result

# Guarda el resultado en outputs/tables/, en vez de la carpeta del cuaderno
export_query_result(
    query_result,
    query_years,
    query_tipo_regimen,
    query_nombre,
    query_ncm,
    output_dir=EXPORT_DIR,
    formato=query_formato,
)
````
