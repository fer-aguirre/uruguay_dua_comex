# Query data

Ad-hoc lookups against `data/processed/dua_2016-2026_decoded.parquet` (built by 0.2), using `uruguay_dua_comex/query.py`.

## Filters (`query_dua`)

- `years`: int or list of years.
- `tipo_regimen` *(optional)*: `"i"` / `"e"` / `"t"` (Importación/Exportación/Tránsito), or a list — case-insensitive.
- `nombre` *(optional)*: importer/exporter name, case-insensitive substring. More specific is better — a bare `"sa"` can match thousands of names.
- `ncm` *(optional)*: code or prefix, or a list. Dots are optional (`"78.01"` == `"7801"`).

At least one of `nombre`/`ncm` is required, so a query can't accidentally scan the full ~20M-row table.

## Export (`export_query_result`)

- Writes to `<output_dir>/<params>_<years>.<formato>` — skipped if the result is empty.
- `formato`: `"csv"` (default), `"xlsx"`, or `"parquet"`.
- `output_dir`: defaults to the current directory if left `None` (created automatically if missing). This notebook passes `EXPORT_DIR = utils.outputs_tables_dir()` so exports land in `outputs/tables/` instead.

## From a terminal

```
python -m uruguay_dua_comex.query --years 2024 --ncm 2701 -r i -f xlsx
```

Same defaults apply (cwd, `--output`/`-o` to override).


````python
from uruguay_dua_comex import utils
from uruguay_dua_comex.query import export_query_result, query_dua

EXPORT_DIR = utils.outputs_tables_dir()
````

````python
# --- Configuration ---
query_years = [2022, 2023, 2024, 2025, 2026]
query_tipo_regimen = None  # e.g. "i", "e", "t"
query_nombre = None
query_ncm = "7801"
query_formato = "csv"  # "csv" (default), "xlsx" or "parquet"

query_result = query_dua(
    years=query_years, tipo_regimen=query_tipo_regimen, nombre=query_nombre, ncm=query_ncm
)
query_result

# Exports under outputs/tables/ rather than the notebook's own directory
export_query_result(
    query_result,
    query_years,
    query_tipo_regimen,
    query_nombre,
    query_ncm,
    output_dir=EXPORT_DIR,
    formato=query_formato,
)
````
